```python
import pandas as pd 
import sqlite3
```

```python
sales = pd.read_csv('data/sales_clean.csv', parse_dates=['InvoiceDate'])
```

```python
conn = sqlite3.connect('data/retail.db')
sales.to_sql('sales', conn, index=False, if_exists='replace')
```

```python
query = "SELECT * FROM sales LIMIT 5"
pd.read_sql(query, conn)
```

```python
query = "SELECT SUM(Revenue) FROM sales"
pd.read_sql(query,conn)
```

```python
sales['Revenue'].sum()
```

```python
query = """
SELECT Country, SUM(revenue) AS Revenue
FROM sales
WHERE Country != 'United Kingdom'
GROUP BY Country 
ORDER BY Revenue DESC
LIMIT 10
"""
pd.read_sql(query,conn)
```

```python
query = """
SELECT Country, SUM(Revenue) AS Revenue, 
COUNT(DISTINCT "Customer ID") AS TotalCustomers
FROM sales
WHERE Country != 'United Kingdom'
GROUP BY Country 
ORDER BY Revenue DESC
LIMIT 10 
"""
pd.read_sql(query,conn)
```

```python
query = """
SELECT StockCode,SUM(Revenue)as Revenue
FROM sales
WHERE StockCode GLOB '[0-9]*'
GROUP BY StockCode
ORDER BY Revenue DESC
LIMIT 10 
"""
pd.read_sql(query,conn)
```

```python
query= """
SELECT strftime ('%Y-%m', InvoiceDate) AS Month,
SUM(Revenue)AS Revenue
FROM sales 
WHERE InvoiceDate < '2011-12-01'
GROUP BY Month 
ORDER BY Revenue DESC
"""
pd.read_sql(query,conn)
```
